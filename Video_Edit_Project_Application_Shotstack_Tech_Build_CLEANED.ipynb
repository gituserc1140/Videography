```python
import requests, time, json

API_KEY = "GVwngZlVHrvu0N2Acbtu1H4cMUiVnFgUJeg6pNLu"
API_URL = "https://api.shotstack.io/stage/render"

def make_tiktok(video_url, caption_text, duration=10.0, caption_duration=5.0, effect="zoomIn"):
    headers = {
        "x-api-key": API_KEY,
        "Content-Type": "application/json"
    }

    edit_payload = {
        "timeline": {
            "background": "#000000",
            "tracks": [
                {
                    "clips": [
                        {
                            "asset": {
                                "type": "video",
                                "src": video_url
                            },
                            "start": 0.0,
                            "length": duration,
                            "fit": "cover",
                            "effect": effect
                        },
                        {
                            "asset": {
                                "type": "title",
                                "text": caption_text,
                                "style": "minimal",
                                "size": "medium",
                                "color": "#FFFFFF"
                            },
                            "start": 0.0,
                            "length": caption_duration,
                            "position": "bottom"
                        }
                    ]
                }
            ]
        },
        "output": {
            "format": "mp4",
            "resolution": "1080",
            "aspectRatio": "9:16"
        }
    }

    # Submit render
    response = requests.post(API_URL, headers=headers, data=json.dumps(edit_payload))
    render_id = response.json()["response"]["id"]
    print("Render ID:", render_id)

    # Poll for completion
    render_url = None
    status = None

    while status not in ["done", "failed"]:
        poll = requests.get(f"{API_URL}/{render_id}", headers=headers)
        status = poll.json()["response"]["status"]
        print("Status:", status)
        if status == "done":
            render_url = poll.json()["response"]["url"]
            print("Render complete:", render_url)
        elif status == "failed":
            print("Render failed:", poll.json()["response"].get("error"))
            return

        time.sleep(5)

    # Download video
    if render_url:
        video = requests.get(render_url)
        with open("tiktok_edit.mp4", "wb") as f:
            f.write(video.content)
        print("Downloaded as tiktok_edit.mp4")
```

```python
punch_segments = [
    {"start": 0.0, "length": 3.0, "effect": "zoomIn"},
    {"start": 3.0, "length": 3.0, "effect": "zoomOut"},
    {"start": 6.0, "length": 3.0, "effect": "slideLeft"}
]
```

```python
def build_punch_clips(video_url, segments):
    clips = []
    timeline_start = 0.0
    for seg in segments:
        clips.append({
            "asset": {
                "type": "video",
                "src": video_url,
                "trim": seg["start"]
            },
            "start": timeline_start,
            "length": seg["length"],
            "fit": "cover",
            "effect": seg["effect"]
        })
        timeline_start += seg["length"]
    return clips
```

```python
def make_tiktok_punch(video_url, segments, caption_text=""):
    headers = {
        "x-api-key": API_KEY,
        "Content-Type": "application/json"
    }

    # Build punch-in clips
    punch_clips = build_punch_clips(video_url, segments)

    # Optional caption
    caption_clip = {
        "asset": {
            "type": "title",
            "text": caption_text,
            "style": "minimal",
            "size": "medium",
            "color": "#FFFFFF"
        },
        "start": 0.0,
        "length": 4.0,
        "position": "bottom"
    } if caption_text else None

    # Build timeline
    track_clips = punch_clips + ([caption_clip] if caption_clip else [])

    edit_payload = {
        "timeline": {
            "background": "#000000",
            "tracks": [
                {"clips": track_clips}
            ]
        },
        "output": {
            "format": "mp4",
            "resolution": "1080",
            "aspectRatio": "9:16"
        }
    }

    # Submit render
    response = requests.post(API_URL, headers=headers, data=json.dumps(edit_payload))

    if response.status_code == 200: # Check for successful response
        response_data = response.json()
        render_id = response_data["response"]["id"]
        print("Render ID:", render_id)
    else:
        print(f"API Error: {response.status_code} - {response.text}")
        return # Exit the function if API call failed

    # Poll + download (same as before)
    render_url = None
    status = None

    while status not in ["done", "failed"]:
        poll = requests.get(f"{API_URL}/{render_id}", headers=headers)
        status = poll.json()["response"]["status"]
        print("Status:", status)
        if status == "done":
            render_url = poll.json()["response"]["url"]
            print("Render complete:", render_url)
        elif status == "failed":
            print("Render failed:", poll.json()["response"].get("error"))
            break
        else:
            time.sleep(5)

    if render_url:
        video = requests.get(render_url)
        with open("tiktok_punch_edit.mp4", "wb") as f:
            f.write(video.content)
        print("Downloaded as tiktok_punch_edit.mp4")
```
